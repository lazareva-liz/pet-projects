- Статистика по демографическим признакам

```python
# Основные метрики
print("ДЕМОГРАФИЯ - ОСНОВНЫЕ МЕТРИКИ:")
print(f"Всего студентов: {len(df)}")
print(f"Средний возраст: {df['age'].mean():.1f} лет")
print(f"Медианный возраст: {df['age'].median():.0f} лет")

male_count = df['gender'].sum()  
female_count = len(df) - male_count  
print(f"\nМужчины: {male_count} ({male_count/len(df)*100:.1f}%)")
print(f"Женщины: {female_count} ({female_count/len(df)*100:.1f}%)")

international_count = df['international'].sum()
local_count = len(df) - international_count
print(f"Резиденство: {local_count} местных/{international_count} иностранных "
      f"({local_count/len(df)*100:.1f}%/{international_count/len(df)*100:.1f}%)")
print('-------------')

# Создаем читаемые метки
df['gender_label'] = df['gender'].map({0: 'Female', 1: 'Male'})
df['international_label'] = df['international'].map({0: 'Local', 1: 'International'})

# Таблица успешности
success_table = pd.pivot_table(df, 
                               index=['gender_label', 'international_label'],
                               columns='target',
                               aggfunc='size',
                               fill_value=0)
print("\nУспешность по демографическим группам:")
success_table
```
